```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "google/gemma-3-1b-it",
    token = ""
)
print("Loaded Successfully")

text = "What is Java"

tokens = tokenizer.tokenize(text)
print(tokens)

encoded = tokenizer(text)
print(encoded)
```
