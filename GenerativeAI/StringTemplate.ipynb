```python
from google import genai

client = genai.Client(
    api_key=""
)

print("Gemini Connected Successfully.")
```

```python
prompt_template = """
{priming}

{question}

{decorator}

your solution:
"""
```

```python
priming_text = "You are an expert Java developer."

question = "Create a Stack implementation in Java."

decorator = "Insert comments for each line of code."
```

```python
prompt = prompt_template.format(
    priming=priming_text,
    question=question,
    decorator=decorator
)

print(prompt)
```

```python

response = client.models.generate_content(
    model = "gemini-3.5-flash",
    contents=prompt
)

print(response.text)
```
