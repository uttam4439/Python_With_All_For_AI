```python
def generate_text(prompt):
    response = client.models.generate_content(
        model = "gemini-3.5-flash",
    contents = prompt
    )

    return response.text
```

```python
answer = generate_text("Write code in java to print Hello world.")
print(answer)
```

```python
#Use of Temperature

from google.genai import types

response = client.models.generate_content(
    model = "gemini-3.5-flash",
    contents = "write a short story on enviornment.",
    config=types.GenerateContentConfig(
        temperature = 1.5   # I have changed temp. from 0 to 1.5 and i got diff. result because less temp. 
                            #less creative and more consistent. (range upto only 0.0-2.0)
    )
)

print(response.text)
```
